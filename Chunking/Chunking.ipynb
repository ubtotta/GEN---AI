# What is chunking?

- The process of converting the larger piece of data into smaller meaning full piece of data is known as chunking and this chunking is used before embedding in RAG archeitecture

# Fixed Size chunking

```python
with open('data_chunk.txt','r') as file:
    data=file.read()
data
```

```python
# fixed_size=20
# chunks=[]
# i=0
# while i<(len(data)):
#     chunk=data[i:fixed_size]
#     chunks.append(chunk)
#     fixed_size+=20
#     i+=20
```

```python
chunks=[]
fixed_size=20
for index in range(0,len(data), fixed_size):
    chunk = data[index:index+fixed_size]
    # print(chunk)
    chunks.append(chunk)
```

```python
chunks
```

# Fixed size + Overlapping chunking

- we can preserve the meaning by using this overlapping
- we can join first chunk with another chunk
- this we can use in RAG
- 20% of fixed size chunking

```python
chunks=[]
fixed_size=20
over_lap=4
step=fixed_size-over_lap
for index in range(0,len(data), step):
    chunk = data[index:index+fixed_size]
    # print(chunk)
    chunks.append(chunk)
```
