```python
with open('data_chunk.txt','r') as file:
    data=file.read()
data
```

```python
fixed_size=20
chunks=[]
i=0
while i<(len(data)):
    chunk=data[i:fixed_size]
    chunks.append(chunk)
    fixed_size+=20
    i+=20
```

```python
chunks=[]
fixed_size=20
for index in range(0,len(data), fixed_size):
    chunk = data[index:index+fixed_size]
    # print(chunk)
    chunks.append(chunk)
```
