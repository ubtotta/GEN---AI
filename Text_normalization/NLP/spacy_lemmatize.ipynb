```python
import spacy

nlp=spacy.load('en_core_web_sm')
```

```python
tokens = nlp('''Package name or model path. vocab A Vocab object. If True, a vocab is created. disable''')
print(type(tokens),tokens)
# internally converts it into tokens , here tokenization is not required , by default it converts ,returning the group of tokens
```

```python
lemmatize_tokens= [token.lemma_ for token in tokens if not token.is_stop]
# for token in tokens:
#     # print(type(token))
#     print(token.lemma_,' ', token.is_stop)

print(lemmatize_tokens)
print(' '.join(lemmatize_tokens))
```


***Advantages***
- internally converts it into tokens , here tokenization is not required , by default it converts
- Converts each character to lower case by itself
- then applies lemmatize using pos tag by default we dont need to mention manually, it recognizes by itself and converts
