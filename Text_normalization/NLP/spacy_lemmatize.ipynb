```python
import spacy

nlp=spacy.load('en_core_web_sm')
```

```python
tokens = nlp('''Package name or model path. vocab A Vocab object. If True, a vocab is created. disable''')
print(type(tokens),tokens)
# internally converts it into tokens , here tokenization is not required , by default it converts ,returning the group of tokens
```

```python
lemmatize_tokens= [token.lemma_ for token in tokens if not token.is_stop]
# for token in tokens:
#     # print(type(token))
#     print(token.lemma_,' ', token.is_stop)

print(lemmatize_tokens)
print(' '.join(lemmatize_tokens))
```


***Advantages***
- internally converts it into tokens , here tokenization is not required , by default it converts
- Converts each character to lower case by itself
- then applies lemmatize using pos tag by default we dont need to mention manually, it recognizes by itself and converts
- easily we can remove the stop words from the tokens
- here we can get name entity recognization
- when it comes to real time projects we use this space its very faster than nltk, in space we have more features compared to nltk

```python
text = ''' PYTHON  I am Robert Downey Jr travelling from Bengaluru to USA, Infosys https://github.com/  flight price is 5,000 in 2004 pyspiders@gmail.com.'''
tokens = nlp(text)
tokens
```

```python
# tokens.ents #real world entities
# print(type(tokens.ents))
for ent in tokens.ents:
    print(ent.text, ent.label_)
```

```python
for token in tokens:
    print(token.pos_, token.tag_)
```
