```python
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
res='''Hello i am udaya kumar\n. 
      I am currenlty doing the intership at Pyspiders in the domain of Python with Data Science. Located at BTM layout, Bangalore.'''
```

```python
words_tokens=word_tokenize(res)
```

```python
words_tokens
```

```python
sent_tokens=sent_tokenize(res)
```

```python
sent_tokens
```

```python
# re.findall(r'\s+',res)
re.sub(r'\s+','',res)
```
