```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
words = [
    "running", "jumps", "easily", "flying", "happiness",
    "played", "singing", "players", "creation", "creative",
    "studies", "studying", "connection", "connected", "connections",
    "running", "runner", "emojies", "dogs", "children","a", "and", "the"
]

for word in words:
    print(f'{word} ---> {stemmer.stem(word)}')
```

- this stemming works based on some rules
- first converts all the chars to lower case
- it just removes the suffix based on rule

```python
from nltk.stem import SnowballStemmer

snowball_Stemmer=SnowballStemmer('english',ignore_stopwords=True)
```

```python
snowball_words = [
    "running", "jumps", "easily", "flying", "happiness",
    "played", "singing", "players", "creation", "creative",
    "studies", "studying", "connection", "connected", "connections",
    "running", "runner", "emojies", "dogs", "children","a", "and", "the"
]

for word in snowball_words:
    print(f'{word} ---> {snowball_Stemmer.stem(word)}')
```

- ***porterstemmer*** : only english language, converts all chars to lower case, includes stopwords
- ***snowballstemmer*** : supports many languages, keeps same case, excludes the stopwords
