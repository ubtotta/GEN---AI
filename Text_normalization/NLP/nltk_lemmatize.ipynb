```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
```

```python
tokens = ['running', 'jumps', 'children', 'better', 'geese', 'cars', 'wolves', 'going', 'ate', 'happier', 'feet', 'houses']
tokens
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
for word in tokens:
    print(word,'--->',lemmatizer.lemmatize(word,pos='v'))
```

- here every word is not being lemmatize because here by default it is using noun
- so here it uses lemmatize method only when words are noun
- here we also need to change the pos stack accordingly how and which type of words are present which is not possible to check manually thats the disadvantage in lemmatization, this we can overcome in spaCy Lemmatize
