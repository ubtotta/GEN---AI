```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger') #this will help in finding word belongs to which pos
nltk.download('averaged_perceptron_tagger_eng') 
```

```python
tokens = ['running', 'jumps', 'children', 'better', 'geese', 'cars', 'wolves', 'going', 'ate', 'happier', 'feet', 'houses']
tokens
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
for word in tokens:
    print(word,'--->',lemmatizer.lemmatize(word,pos='v'))
```

### Lemmatizer
- here every word is not being lemmatize because here by default it is using noun
- so here it uses lemmatize method only when words are noun
- here we also need to change the pos stack accordingly how and which type of words are present which is not possible to check manually thats the disadvantage in lemmatization, this we can overcome in spaCy Lemmatize

### Difference between stemming and lemmatize
- Stemming works based on set of rules, directly removes the suffix wont check that if a word is noun or pronoun etc
- Lemmatize works based on pos(parts of speech) stack, suppose if we mention as pos='noun' then it will check is the word noun if it is noun then it applies lemmatization for that word

***The Part Of Speech tag. Valid options are "n" for nouns, "v" for verbs, "a" for adjectives, "r" for adverbs and "s" for satellite adjectives.***

```python
updated_tokens=nltk.pos_tag(tokens)
```

```python
''' 
N - 'n' 
V - 'v'
JJ - 'a'(adjectives)
R - 'r'(adverb)
'n'
'''
```

```python
def find_postag(tag):
    if tag.startswith('J'):
        return 'a'
    elif tag.startswith('N'):
        return 'n'
    elif tag.startswith('V'):
        return 'v'
    elif tag.startswith('R'):
        return 'r'
    else:
        return 'n'
```

```python
updated_tokens=dict(updated_tokens)
```

```python
for k,v in updated_tokens.items():
    print(k,'--->',lemmatizer.lemmatize(k,pos=find_postag(v)))
```
