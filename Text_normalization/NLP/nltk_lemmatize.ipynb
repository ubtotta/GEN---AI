```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
```

```python
tokens = ['running', 'jumps', 'children', 'better', 'geese', 'cars', 'wolves', 'going', 'ate', 'happier', 'feet']
tokens
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
for word in tokens:
    print(word,'--->',lemmatizer.lemmatize(word))
```

- here every word is not being lemmatize because here by default it is using noun
- so here it uses lemmatize method only when words are noun
