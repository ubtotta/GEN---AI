```python
with open('data.txt', 'r+') as f:
    puncutations='''!"#$%&'()*+,-./:;<=>?@[\]^_`{|}~'''
    data=f.read()
    f.seek(0)
    new_file=''
    for i in data:
        if i not in puncutations:
            new_file+=i
    f.write(new_file)
    f.truncate()
```
